```python
# for saving dataset that is uploaded from local machine
import io
 # io module allows us to manage the file-related input and output operations
#df = pd.read_csv(io.BytesIO(uploaded['tree.csv']))
df = pd.read_csv('tree.csv')
# in [] give the name of the file that you have uploaded, here df is a variable
#.. to store data, pd is panda library which we imported first, read_csv is a
#.. function of panda that will read csv
# file, io.BytesIO is a method to convert a byte data into file/object data
print(df)
# this will will simple print the data
```

```python
#data = pd.read_csv('/content/drive/Shareddrives/Research_datasets/tree.csv') #inserting data from google drive, here "data" is a variable to store data
data = df.drop('Days', axis=1)
# drop one column name as Days
data
 # show entire dataset
```

```python
data.isnull().sum()
 #checking is there any null value
```

```python
y = data['Wear Jacket?']
# dividing dataset into y
y
 # show resultant data
```

```python
x = data.drop('Wear Jacket?', axis=1)
# dividing dataset into y
x
# show attribute data
```

```python
from sklearn.preprocessing import LabelEncoder
 # importing laben encoding for converting string into unique number
```

```python
# data encoding from text to number
le = LabelEncoder()
# taking instance of LabelEncoder() class
# The fit(data) method is used to compute the mean and std dev for a given feature to be
# used further for scaling. The transform(data) method is used to perform scaling using
# mean and std dev calculated using the . fit() method. The fit_transform() method does both fits and transform.
x.Outlook = le.fit_transform(data['Outlook'])
# encoding Outlook column
x.Temprature = le.fit_transform(data['Temprature'])
# encoding Outlook column
x.Routine = le.fit_transform(data['Routine'])
# encoding Outlook column
x
```

```python
#Import Gaussian Naive Bayes model
# GaussianNB to implement the Gaussian Naïve Bayes algorithm for classification
from sklearn.naive_bayes import GaussianNB
# sklearn.naive_bayes is a module in the scikit-learn library that provides implementations of the Naive Bayes algorithm for classification and prediction tasks.
# Gaussian Naive Bayes function, i.e. GaussianNB()
#scikit-learn is a free software machine learning
#..library for the Python programming language
#Create a Gaussian Classifier that is a class that have methods/functions like fit, predict, score etc.
model = GaussianNB()
# Train the model using the training sets
model.fit(x,y)
#Predict Output
# Python predict() function enables us to predict the labels of the data values
# on the basis of the trained model. The predict() function accepts only a single argument which is usually the data to be tested.
predicted= model.predict([[1, 1, 0]])
# 1=sunny; 1=warm; 0=indoor predicting the result
print("Predicted Value:", predicted)
# output: Predicted Value: ['No']
```
